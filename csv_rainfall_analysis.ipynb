```python
import pandas as pd

data = {
    "Crop": ["Maize", "Rice", "Bean", "Tomato", "Yam"],
    "Yield_kg": [1500, 2000, 800, 1200, 3000],
    "Price_Naira": [350, 400, 800, 500, 600]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.describe())
```

```python
df.to_csv("nigerian_crops.csv", index=False)
print("CVS file saved successfully!")

new_df = pd.read_csv("nigerian_crops.csv")
print(new_df)
```
